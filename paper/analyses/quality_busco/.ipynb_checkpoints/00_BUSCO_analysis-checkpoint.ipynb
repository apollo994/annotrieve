```python
import io
import requests
import pandas as pd
import matplotlib.pyplot as plt

API = "https://genome.crg.es/annotrieve/api/v0"

# taxonomic groups of interest (taxids resolved with the Annotrieve /taxons endpoint)
GROUPS = {
    "Mammalia": 40674,
    "Primates": 9443,
    "Solanaceae": 4070,
    "Papilionoideae": 3814,
    "Saccharomyces": 4930,
}
```

```python
# use annotrieve API to check the annotation counts of the selected groups
rows = []
for name, taxid in GROUPS.items():
    t = requests.get(f"{API}/taxons/{taxid}", timeout=60)
    t.raise_for_status()
    t = t.json()
    rows.append((taxid, t["scientific_name"], t["rank"], t["annotations_count"]))

groups = pd.DataFrame(rows, columns=["taxid", "taxon_name", "rank", "annotations_count"])
groups
```

```python
# get one annotation per species (the one with the highest BUSCO complete) for each group
def fetch_busco(taxid):
    """All annotations of a taxon that have a BUSCO complete value."""
    body = {
        "taxids": str(taxid),
        "busco_complete_from": 0,
        "selected_fields": "busco_complete,busco_lineage",
    }
    resp = requests.post(f"{API}/annotations/report", json=body, timeout=300)
    resp.raise_for_status()
    df = pd.read_csv(io.StringIO(resp.text), sep="\t")
    return df.dropna(subset=["busco_complete"])

selected = []
for _, row in groups.iterrows():
    df = fetch_busco(row["taxid"])
    n_annotations = len(df)
    # species = genus + epithet, so strains/isolates with their own taxid collapse into their species
    df = df.assign(species=df["organism_name"].str.split().str[:2].str.join(" "))
    df = (
        df.sort_values("busco_complete", ascending=False)
        .drop_duplicates("species")
        .assign(group=row["taxon_name"])
    )
    print(f"{row['taxon_name']}: {len(df)} species ({n_annotations} annotations with BUSCO "
          f"of {row['annotations_count']})")
    selected.append(df)

busco = pd.concat(selected, ignore_index=True)
busco[["group", "species", "organism_name", "taxid", "annotation_id", "busco_complete"]].head()
```

```python
# plot complete percentage BUSCO values distribution in 5 panes using histograms, one per group with one value per species
fig, axes = plt.subplots(1, len(groups), figsize=(4 * len(groups), 3.5), sharex=True)
bins = range(0, 101, 5)

for ax, group in zip(axes, groups["taxon_name"]):
    vals = busco.loc[busco["group"] == group, "busco_complete"]
    ax.hist(vals, bins=bins, edgecolor="white")
    ax.set_title(f"{group} (n={len(vals)})")
    ax.set_xlabel("BUSCO complete (%)")

axes[0].set_ylabel("Number of species")
fig.tight_layout()
plt.show()
```
